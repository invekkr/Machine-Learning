# Python Prerequisites: 08. Core ML Programming Patterns

Practical implementation of universal ML programming conventions: (N, D) shapes, hyperparameters vs parameters, and reproducibility.

```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# 1. The (N, D) Convention
# N = Number of samples (rows), D = Number of features (columns)
X = np.array([
    [25, 50000],  # Sample 0
    [30, 70000],  # Sample 1
    [45, 120000], # Sample 2
    [35, 85000]   # Sample 3
])
y = np.array([0, 1, 1, 0])

N, D = X.shape
print(f"N (Samples):  {N}")
print(f"D (Features): {D}")
```

```python
# 2. Hyperparameters vs. Parameters
# Hyperparameter: n_neighbors=3 set by the ENGINEER before training
knn = KNeighborsClassifier(n_neighbors=3)

# Training: Algorithm computes internal parameters
knn.fit(X, y)

# Predictions on new data
new_sample = np.array([[28, 60000]])
pred = knn.predict(new_sample)
print(f"Prediction for customer with Age=28, Salary=$60,000: {pred[0]}")
```

```python
# 3. Reproducibility via random seed
np.random.seed(42)
shuffled_indices = np.random.permutation(len(X))
print("Reproducible random permutation:", shuffled_indices)
```
