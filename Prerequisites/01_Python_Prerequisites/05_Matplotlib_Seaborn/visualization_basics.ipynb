# Python Prerequisites: 05. Matplotlib & Seaborn Basics

Practical implementation of essential visualization plots for Exploratory Data Analysis.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Sample dataset
np.random.seed(42)
df = pd.DataFrame({
    'Age': np.random.randint(20, 60, size=30),
    'Salary': np.random.randint(30000, 120000, size=30),
    'Gender': np.random.choice(['Male', 'Female'], size=30),
    'Purchased': np.random.choice([0, 1], size=30)
})

sns.set_theme(style='whitegrid')
print("Sample Data for Visualization:")
display(df.head(4))
```

```python
# 1. Scatter Plot (Numerical vs Numerical)
plt.figure(figsize=(6, 4))
sns.scatterplot(x='Age', y='Salary', hue='Purchased', data=df, palette='Set1', s=80)
plt.title("Scatter Plot: Age vs. Salary")
plt.show()
```

```python
# 2. Histogram with KDE (Distribution)
plt.figure(figsize=(6, 4))
sns.histplot(df['Age'], kde=True, bins=6, color='teal')
plt.title("Distribution of Age")
plt.show()
```

```python
# 3. Bar Plot (Categorical Counts)
plt.figure(figsize=(5, 3))
sns.countplot(x='Gender', data=df, palette='pastel')
plt.title("Gender Frequency Counts")
plt.show()
```

```python
# 4. Box Plot (Spread & Outliers)
plt.figure(figsize=(6, 4))
sns.boxplot(x='Purchased', y='Salary', data=df, palette='Blues')
plt.title("Salary Distribution by Purchase Status")
plt.show()
```

```python
# 5. Correlation Heatmap
plt.figure(figsize=(5, 4))
corr = df[['Age', 'Salary', 'Purchased']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title("Feature Correlation Heatmap")
plt.show()
```
