# Python Prerequisites: 01. Python Basics

Practical implementation of core Python data types, structures, loops, and functions for Machine Learning.

```python
# 1. Variables and Data Types
age = 25              # int
salary = 55000.50     # float
city = "Delhi"        # str
is_student = False    # bool
unknown_val = None    # NoneType

print(f"age: {age} ({type(age)})")
print(f"salary: {salary} ({type(salary)})")
print(f"city: {city} ({type(city)})")
print(f"is_student: {is_student} ({type(is_student)})")
print(f"unknown_val: {unknown_val} ({type(unknown_val)})")
```

```python
# 2. Collections: Lists, Tuples, Sets, Dictionaries
# List: Ordered and mutable
feature_names = ['age', 'salary', 'experience']
feature_names.append('credit_score')
print("List:", feature_names)

# Tuple: Ordered and immutable
fixed_shape = (28, 28)
print("Tuple:", fixed_shape)

# Set: Unique elements (duplicates removed)
categories = ['cat', 'dog', 'cat', 'bird', 'dog']
unique_cats = set(categories)
print("Set:", unique_cats)

# Dictionary: Key-value mapping
sample_record = {'age': 30, 'salary': 70000, 'approved': 1}
print("Dictionary:", sample_record)
print("Accessing salary:", sample_record['salary'])
```

```python
# 3. Indexing and Slicing
numbers = [10, 20, 30, 40, 50]
print("First element [0]:", numbers[0])
print("Last element [-1]:", numbers[-1])
print("Slice [1:4]:     ", numbers[1:4])
```

```python
# 4. Control Flow and Loops
# Conditionals
score = 0.85
if score >= 0.80:
    verdict = "High probability"
elif score >= 0.50:
    verdict = "Medium probability"
else:
    verdict = "Low probability"
print("Verdict:", verdict)

# For Loop
raw_errors = [2.0, -1.0, 3.0]
total_abs_error = 0.0
for err in raw_errors:
    total_abs_error += abs(err)
print("Total absolute error:", total_abs_error)

# List Comprehension (Pythonic data transformation)
squared_errors = [err**2 for err in raw_errors]
print("Squared errors:", squared_errors)
```

```python
# 5. Functions
def compute_mean(values):
    """Calculates average of numbers."""
    return sum(values) / len(values)

sample_data = [10, 20, 30, 40]
print(f"Mean of {sample_data}: {compute_mean(sample_data)}")
```
