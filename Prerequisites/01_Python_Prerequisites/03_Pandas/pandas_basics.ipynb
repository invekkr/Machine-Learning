# Python Prerequisites: 03. Pandas Basics

Practical implementation of Pandas Series, DataFrames, indexing, filtering, and missing values.

```python
import pandas as pd
import numpy as np

# 1. Creating Series and DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25.0, 30.0, np.nan, 35.0],
    'Salary': [50000.0, 70000.0, 60000.0, 85000.0],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Bangalore'],
    'Purchased': [0, 1, 0, 1]
})

print("--- DataFrame Inspection ---")
display(df)
print("\nData Types:")
print(df.dtypes)
```

```python
# 2. Selecting Columns and Rows
# Single column (Series)
print("Ages Series:\n", df['Age'])

# Multiple columns (DataFrame)
print("\nSubset DataFrame:")
display(df[['Name', 'Salary']])

# Row indexing via iloc (by position)
print("\nRow 0:", df.iloc[0]['Name'])
```

```python
# 3. Filtering Rows with Boolean Conditions
high_earners = df[df['Salary'] >= 65000]
print("Employees earning >= $65,000:")
display(high_earners)
```

```python
# 4. Missing Value Detection and Imputation
print("Missing values count per column:")
print(df.isnull().sum())

# Fill missing Age with median
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)
print(f"\nFilled missing Age with median ({median_age}):")
display(df)
```

```python
# 5. Summary Statistics and GroupBy
print("--- Numerical Summary ---")
display(df.describe())

print("\nAverage Salary by City:")
print(df.groupby('City')['Salary'].mean())
```
