# Mathematics Prerequisites: 03. Functions & Graphs

Practical implementation of functions, slope, intercept, and coordinate plotting.

```python
# 1. Defining and Evaluating a Linear Function: f(x) = 2x + 1
def f(x):
    return 2 * x + 1

inputs = [0, 1, 2, 3, 4]
outputs = [f(x) for x in inputs]

for inp, out in zip(inputs, outputs):
    print(f"x = {inp} -> f(x) = {out}")
```

```python
# 2. Calculating Slope (m) from Coordinates
# P1 = (1, 3), P2 = (2, 5)
x1, y1 = 1, 3
x2, y2 = 2, 5

slope = (y2 - y1) / (x2 - x1)
intercept = y1 - slope * x1

print(f"Calculated Slope m:     {slope}")
print(f"Calculated Intercept b: {intercept}")
print(f"Equation of line:       y = {slope}x + {intercept}")
```

```python
# 3. Plotting the Linear Function
import matplotlib.pyplot as plt
import numpy as np

x_vals = np.linspace(-1, 5, 50)
y_vals = f(x_vals)

plt.figure(figsize=(6, 4))
plt.plot(x_vals, y_vals, color='green', label='y = 2x + 1')
plt.scatter([1, 2], [3, 5], color='red', s=80, zorder=5, label='Points (1,3) & (2,5)')
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.title("Graph of Linear Function: y = 2x + 1")
plt.xlabel("x (Input)")
plt.ylabel("y (Output)")
plt.legend()
plt.grid(True)
plt.show()
```
