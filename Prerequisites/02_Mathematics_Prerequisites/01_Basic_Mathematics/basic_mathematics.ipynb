# Mathematics Prerequisites: 01. Basic Mathematics

Practical implementation of powers, square roots, logarithms, and scientific notation.

```python
import math
import numpy as np

# 1. Powers and Square Roots
print("2^3 (2 cubed)  =", 2**3)
print("sqrt(9)        =", math.sqrt(9))
print("sqrt(25)       =", math.sqrt(25))
```

```python
# 2. Logarithms (Demystified!)
# 10^2 = 100 -> log10(100) = 2
print("log10(10)   =", math.log10(10))
print("log10(100)  =", math.log10(100))
print("log10(1000) =", math.log10(1000))

# 2^3 = 8 -> log2(8) = 3 (Used in Decision Trees / Entropy)
print("\nlog2(2)     =", math.log2(2))
print("log2(4)     =", math.log2(4))
print("log2(8)     =", math.log2(8))
```

```python
# 3. Euler's Number (e) and Natural Log (ln)
# Used in Logistic Regression and Neural Networks
print("e (Euler's number) =", math.e)
print("ln(e)              =", math.log(math.e))
print("ln(e^2)            =", math.log(math.e**2))
```
