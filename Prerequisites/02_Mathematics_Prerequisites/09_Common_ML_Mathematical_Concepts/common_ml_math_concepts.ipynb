# Mathematics Prerequisites: 09. Common ML Mathematical Concepts

Practical implementation of Euclidean/Manhattan distance, MAE/MSE, Sigmoid, and Entropy.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Distance Metrics: Euclidean vs. Manhattan
# P1 = (1, 2), P2 = (4, 6)
p1 = np.array([1, 2])
p2 = np.array([4, 6])

euclidean_dist = np.sqrt(np.sum((p1 - p2)**2))
manhattan_dist = np.sum(np.abs(p1 - p2))

print("Points: (1, 2) and (4, 6)")
print(f"Euclidean Distance (Straight line): {euclidean_dist:.1f}")
print(f"Manhattan Distance (City grid):    {manhattan_dist:.1f}")
```

```python
# 2. Regression Error Metrics (MAE, MSE, RMSE)
actual = np.array([10.0, 20.0, 30.0])
predicted = np.array([12.0, 18.0, 31.0])

errors = actual - predicted
mae = np.mean(np.abs(errors))
mse = np.mean(errors**2)
rmse = np.sqrt(mse)

print("Actual:   ", actual)
print("Predicted:", predicted)
print("Residuals:", errors)
print(f"MAE:  {mae:.2f}")
print(f"MSE:  {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
```

```python
# 3. The Sigmoid Function (Foundation of Logistic Regression)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z_vals = np.linspace(-6, 6, 100)
probabilities = sigmoid(z_vals)

plt.figure(figsize=(6, 4))
plt.plot(z_vals, probabilities, color='red', lw=2)
plt.axhline(0.5, color='gray', linestyle='--', label='Decision Threshold (0.5)')
plt.axvline(0, color='gray', linestyle='--')
plt.title("Sigmoid Function: Squashing Real Numbers into [0, 1]")
plt.xlabel("Raw Linear Score (z)")
plt.ylabel("Probability P(y=1)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 4. Entropy (Foundation of Decision Trees)
# Pure group (100% Yes, 0% No) vs. Mixed group (50% Yes, 50% No)
def binary_entropy(p_positive):
    if p_positive == 0 or p_positive == 1:
        return 0.0
    p_negative = 1 - p_positive
    return -(p_positive * np.log2(p_positive) + p_negative * np.log2(p_negative))

print("Entropy of Pure Group (p=1.0):   ", binary_entropy(1.0), "(Zero uncertainty!)")
print("Entropy of Balanced Group (p=0.5):", binary_entropy(0.5), "(Maximum uncertainty!)")
print("Entropy of Moderately Mixed (p=0.8):", round(binary_entropy(0.8), 3))
```
