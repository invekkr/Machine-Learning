# Mathematics Prerequisites: 08. Optimization & Gradient Descent

Practical implementation of a 1D Gradient Descent optimization loop from scratch.

```python
# 1. 1D Gradient Descent from Scratch
# Objective: Find weight 'w' that minimizes Loss(w) = w^2
# Gradient: d(Loss)/dw = 2w
# Update rule: w = w - learning_rate * gradient

w = 5.0              # Starting guess (high error!)
learning_rate = 0.1  # Step size (alpha)
epochs = 20          # Number of optimization steps

print(f"Starting at w = {w:.4f} (Initial Loss = {w**2:.4f})\n")

history_w = [w]
history_loss = [w**2]

for step in range(1, epochs + 1):
    gradient = 2 * w            # Compute derivative
    w = w - learning_rate * gradient  # Take step downhill!
    loss = w**2
    history_w.append(w)
    history_loss.append(loss)
    if step % 5 == 0 or step == 1:
        print(f"Step {step:2d}: w = {w:.5f} | Gradient = {gradient:.5f} | Loss = {loss:.5f}")

print(f"\nOptimization complete! Final w: {w:.5f} (Target minimum is 0.0)")
```

```python
# 2. Visualizing Convergence
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.plot(range(len(history_loss)), history_loss, marker='o', color='purple')
plt.title("Gradient Descent: Loss Reduction Over Steps")
plt.xlabel("Step / Iteration")
plt.ylabel("Loss (Error)")
plt.grid(True)
plt.show()
```
