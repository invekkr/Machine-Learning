Step-by-step manual calculations and Python verification for Central Tendency and Spread.

```python
import numpy as np

# 1. Dataset: 10, 20, 30
data = np.array([10, 20, 30])

# Manual Mean calculation: (10 + 20 + 30) / 3 = 20
manual_mean = sum(data) / len(data)
print(f"Manual Mean: {manual_mean} | NumPy Mean: {np.mean(data)}")
```

```python
# 2. Manual Variance and Standard Deviation
# Mean = 20
# Differences: (10-20)=-10, (20-20)=0, (30-20)=10
# Squared diffs: 100, 0, 100
# Variance: (100 + 0 + 100) / 3 = 200 / 3 = 66.67
# Std: sqrt(66.67) = 8.16
diffs = data - manual_mean
sq_diffs = diffs ** 2
manual_variance = sum(sq_diffs) / len(data)
manual_std = manual_variance ** 0.5

print(f"Manual Variance: {manual_variance:.2f} | NumPy Variance: {np.var(data):.2f}")
print(f"Manual Std Dev:  {manual_std:.2f}  | NumPy Std Dev:  {np.std(data):.2f}")
```

```python
# 3. Z-Score Standardization (z = (x - mu) / sigma)
z_scores = (data - manual_mean) / manual_std
print("Original Values:     ", data)
print("Standardized Z-Scores:", np.round(z_scores, 2))
print("Mean of Z-Scores:    ", round(np.mean(z_scores), 2))
print("Std of Z-Scores:     ", round(np.std(z_scores), 2))
```
