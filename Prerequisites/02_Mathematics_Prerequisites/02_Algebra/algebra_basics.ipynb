Practical implementation of solving equations and visualizing quadratic error bowls.

```python
# 1. Solving Linear Equations Step-by-Step
# Problem: 3x + 4 = 19
# Step 1: 3x = 19 - 4 = 15
# Step 2: x = 15 / 3 = 5
c = 19
intercept = 4
slope = 3
x = (c - intercept) / slope
print(f"Solution to 3x + 4 = 19 is: x = {x}")
```

```python
# 2. Visualizing a Quadratic Loss Curve (Parabola)
import matplotlib.pyplot as plt
import numpy as np

# y = w^2 (representing error bowl in Machine Learning)
w = np.linspace(-4, 4, 100)
loss = w**2

plt.figure(figsize=(6, 4))
plt.plot(w, loss, color='blue', lw=2)
plt.scatter([0], [0], color='red', s=100, zorder=5, label='Global Minimum (w=0, Loss=0)')
plt.title("Parabolic Loss Curve (Quadratic Function)")
plt.xlabel("Model Weight (w)")
plt.ylabel("Loss (Error)")
plt.legend()
plt.grid(True)
plt.show()
```
