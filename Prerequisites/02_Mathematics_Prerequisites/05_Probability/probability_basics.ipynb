# Mathematics Prerequisites: 05. Probability & Bayes' Theorem

Practical implementation of basic probability, conditional probability, and Bayes' Theorem.

```python
# 1. Classical Probability (Coin Toss & Die Roll)
# Coin: P(Heads) = 1/2
p_heads = 1 / 2
print(f"Probability of Heads: {p_heads * 100:.1f}%")

# 6-sided die: Even numbers {2, 4, 6} out of {1, 2, 3, 4, 5, 6}
p_even = 3 / 6
print(f"Probability of Even Die Roll: {p_even * 100:.1f}%")
```

```python
# 2. Independent Events (Multiplication Rule)
# Probability of rolling two 6s in a row: (1/6) * (1/6)
p_two_sixes = (1/6) * (1/6)
print(f"P(Two Sixes in a row): {p_two_sixes:.4f} ({p_two_sixes * 100:.2f}%)")
```

```python
# 3. Bayes' Theorem Calculation
# P(Spam | 'FREE') = [P('FREE' | Spam) * P(Spam)] / P('FREE')
p_spam = 0.20           # Prior belief: 20% of emails are spam
p_free_given_spam = 0.80 # Likelihood: 80% of spam emails contain 'FREE'
p_free = 0.25           # Evidence: 25% of all emails contain 'FREE'

p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print("--- Bayes' Theorem Email Filter ---")
print(f"Prior Probability P(Spam):           {p_spam * 100:.1f}%")
print(f"Updated Posterior P(Spam | 'FREE'): {p_spam_given_free * 100:.1f}%")
```
