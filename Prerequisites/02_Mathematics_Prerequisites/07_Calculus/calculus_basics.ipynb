Practical implementation of numerical differentiation, slopes, and partial derivatives.

```python
# 1. Numerical Approximation of Derivative (Slope)
# f(x) = x^2
# Power Rule tells us f'(x) = 2x
def f(x):
    return x**2

def numerical_derivative(func, x, h=1e-5):
    """Approximates df/dx as [f(x+h) - f(x)] / h"""
    return (func(x + h) - func(x)) / h

for val in [-2, 0, 3]:
    analytical = 2 * val
    numerical = numerical_derivative(f, val)
    print(f"x = {val:2d} -> Analytical slope (2x): {analytical:2d} | Numerical slope: {numerical:.4f}")
```

```python
# 2. Partial Derivatives and Gradient Vector
# Function: f(x, y) = x^2 + y^2
# df/dx = 2x (treat y as constant)
# df/dy = 2y (treat x as constant)
# Gradient vector = [2x, 2y]

x_pt, y_pt = 3.0, 4.0
grad_x = 2 * x_pt
grad_y = 2 * y_pt

print(f"Point (x={x_pt}, y={y_pt}):")
print(f"Partial derivative w.r.t x: {grad_x}")
print(f"Partial derivative w.r.t y: {grad_y}")
print(f"Gradient vector [df/dx, df/dy]: [{grad_x}, {grad_y}]")
print(f"Uphill direction:   [{grad_x}, {grad_y}]")
print(f"Downhill direction: [{-grad_x}, {-grad_y}] (Used in Gradient Descent!)")
```
